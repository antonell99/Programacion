## Solución

```python
espaniol_ingles = input('Introduce las traduciones(formato: palabra:traduccion, separados por coma): ')
diccionario = {}
for par in espaniol_ingles.split(','):
    espanol, ingles = par.split(':')
    diccionario[espanol.strip()] = ingles.strip()
frase = input('Ingrese frase en espanol: ')
palabras = frase.split()
frase_traducida = [diccionario.get(palabra, palabra) for palabra in palabras]
print("Frase traducida:", " ".join(frase_traducida))
```
