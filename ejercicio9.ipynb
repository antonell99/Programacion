## Ejercicio

Escribir un programa que pregunte al usuario la fecha de su nacimiento en formato `dd/mm/aaaa` y muestra por pantalla, el día, el mes y el año.

## Solución 1

```python
fecha= 'dd/mm/aaaa'
usuario=input('Escribe tu fecha de nacimiento dd/mm/aaaa: ')
dia,mes,ano= usuario.split('/')
print('Dia: ', dia)
print('Mes: ', mes)
print('Ano: ', ano)
```
